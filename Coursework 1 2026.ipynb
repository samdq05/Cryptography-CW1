## Q1 The Vigenère Cipher

Let $\cA = \Z/m\Z$ (for some positive integer $m$) and let $\cP = \cC = \cA^n$ (for some fixed positive integer $n$). Thus $\cP$ and $\cC$ can be seen as the set of length $n$ vectors with entries in $\cA$.

For each key $k \in \cA^n$, let the encryption function $E_k : \cP \to \cC$ be defined by
$$
E_k(b) \equiv b + k \pmod{m},
$$
where the addition is performed component-wise, modulo $m$, on the components of the vectors $b$ and $k$.

This is the **Vigenère cipher** with block length $n$, modulo $m$.

The code below is taken (with slight adaptations) from the file `Week01.ipynb` in the Week 1 ELE tile, where it is referred to as 'Week 1 Code'.

I **strongly recommend** working through the file `Week01.ipynb` before attempting the problems 

The function `str_to_int` represents a string of characters as a numpy array of integers. This isn't encryption or decryption.

```python
def str_to_int(str):
    """Convert a string of characters to a numpy array of integers.
    'A' -> 0, 'B' -> 1, ..., 'Z'-> 26, '[' -> 27, ..., '`' -> 31"""
    start = ord('A')
    f = np.vectorize(lambda x : ord(x)-start)
    return f(list(str))
```

The function `int_to_str` represents an array of integers as a string of characters. This isn't encryption or decryption

```python
def int_to_str(x):
    """Convert a numpy array of integers to a string of characters.
    0 -> 'A', 1 -> 'B', ..., 26 -> 'Z', 27 -> '[', ..., 31 -> '`'"""
    start = ord('A')
    f = np.vectorize(lambda x: chr(x + start))
    return ''.join(f(x))
```

For example, we can convert the numpy array $[20, 28, 31]$ of integers to a string of characters.

```python
int_to_str(np.array([20, 28, 31]))
```

```python
def vigenere_encrypt(k, b, m):
    """The vigenere encryption of vector b by the key k, modulo m"""
    return (k + b) % m
```

In this question, we work modulo $m = 32$ with block length $n = 7$.

## Question 3(a)(i) [10 marks]

Explain how to recover the initialisation vector. Give your answer in the markdown cell below (no code required).

Since we are given two consecutive plaintext and ciphertext blocks, we can derive the following relationships between the values:

$c_1 = (c_0 \oplus b_1) + k \, (\text{mod } 32)$ 

$c_2 = (c_1 \oplus b_2) + k \, (\text{mod } 32)$

If we rearrange each equation to make $k$ the subject and then equate them, that gives the following:

$c_1 - (c_0 \oplus b_1) \, (\text{mod } 32) = c_2 - (c_1 \oplus b_2) \, (\text{mod } 32) $

Now, we can rearrange for the initialisation vector $c_0$

$c_0 \, (\text{mod } 32) = (c_1 + (c_1 \oplus b_2) - c_2) \oplus b_1 \, (\text{mod } 32)$
